```python
import pandas as pd
import csv
import re

file_path = "/content/kaggle_s&p500_2000-2024.csv"

companies = []
tickers = []

with open(file_path, "r", encoding="utf-8-sig") as f:
    lines = f.readlines()

# Skip the malformed header
for line in lines[1:]:
    line = line.strip()

    # Remove the outer quotation marks
    if line.startswith('"') and line.endswith('"'):
        line = line[1:-1]

    # The ticker is always the final item after the last comma
    parts = line.rsplit(",", 1)

    if len(parts) == 2:
        company_part = parts[0]
        ticker = parts[1].strip().strip('"')

        # Remove the index at the beginning
        company_part = re.sub(r'^\d+,', '', company_part)

        # Clean up the doubled quotation marks
        company_part = company_part.replace('""', '"').strip('"').strip()

        companies.append(company_part)
        tickers.append(ticker)

# Create a clean DataFrame
sp500 = pd.DataFrame({
    "Company": companies,
    "Ticker": tickers
})

print("Number of companies:", len(sp500))
print("Columns:", sp500.columns.tolist())

display(sp500.head(10))
```

```python
import requests, time

H = {"User-Agent": "Your Name your.email@university.edu"}   # SEC requires a real contact here

tk = requests.get("https://www.sec.gov/files/company_tickers.json", headers=H).json()
cik_map = {v["ticker"]: str(v["cik_str"]).zfill(10) for v in tk.values()}

def shares_edgar(t, month_index):
    cik = cik_map.get(t) or cik_map.get(t.replace("-", "."))
    if cik is None:
        return None
    r = requests.get(f"https://data.sec.gov/api/xbrl/companyfacts/CIK{cik}.json", headers=H)
    if r.status_code != 200:
        return None
    f = r.json()["facts"]
    rows = []
    for ns, tag in [("dei", "EntityCommonStockSharesOutstanding"),
                    ("us-gaap", "CommonStockSharesOutstanding")]:
        try:
            rows += f[ns][tag]["units"]["shares"]
        except KeyError:
            pass
    if not rows:
        return None
    d = pd.DataFrame(rows)[["filed", "val"]]
    d["filed"] = pd.to_datetime(d["filed"])
    d = d.sort_values("filed").drop_duplicates("filed", keep="last")

    # put old share counts on today's split basis, to match Yahoo's adjusted prices
    sp = yf.Ticker(t).splits
    if len(sp):
        sp.index = sp.index.tz_localize(None)
        d["val"] = d["val"] * d["filed"].apply(lambda x: sp[sp.index > x].prod())

    m = pd.DataFrame({"Date": month_index})
    out = pd.merge_asof(m, d.rename(columns={"filed": "Date"}),
                        on="Date", direction="backward")
    return out.set_index("Date")["val"]

tickers = ["AAPL", "MSFT", "XOM"]          # test first; later: list(vol.columns)
px = yf.download(tickers, start="1999-12-01", end="2025-01-01",
                 auto_adjust=False, progress=False)["Close"].resample("ME").last()

shares, failed = {}, []
for t in tickers:
    s = shares_edgar(t, px.index)
    if s is None: failed.append(t)
    else: shares[t] = s
    time.sleep(0.2)                         # SEC rate limit is 10 requests/second
shares = pd.DataFrame(shares)

mcap = px[shares.columns] * shares
size = np.log(mcap).shift(1)
print("failed:", failed)
print((mcap.dropna(how="all") / 1e9).iloc[::24].round(1))
```

```python
import requests, time, pickle
import yfinance as yf, pandas as pd, numpy as np
from google.colab import files

H = {"User-Agent": "Your Name your.email@university.edu"}   # SEC requires real contact info
N = None                                 # test with 10 tickers; set to None for all
tickers = list(vol.columns)[:N]
months = pd.DatetimeIndex(vol.index).astype("datetime64[ns]")

# ---------- prices (split-adjusted month-end close) ----------
px = yf.download(tickers, start="1999-12-01", end="2025-01-01",
                 auto_adjust=False, progress=False)["Close"]
px = px.resample("ME").last().reindex(months)

# ---------- ticker -> CIK ----------
tk = requests.get("https://www.sec.gov/files/company_tickers.json", headers=H).json()
cik_map = {v["ticker"]: str(v["cik_str"]).zfill(10) for v in tk.values()}

# Priority order. Weighted-average shares is a weak last resort (period average).
TAGS = [("dei", "EntityCommonStockSharesOutstanding"),
        ("us-gaap", "CommonStockSharesOutstanding"),
        ("us-gaap", "WeightedAverageNumberOfSharesOutstandingBasic")]

def splits_for(t):
    sp = yf.Ticker(t).splits
    if len(sp) and sp.index.tz is not None:
        sp.index = sp.index.tz_localize(None)
    return sp

def split_adjust(d, sp):
    """Put old share counts on today's split basis, matching Yahoo's adjusted prices."""
    if len(sp):
        d = d.copy()
        d["val"] = d["val"] * d["filed"].apply(lambda x: sp[sp.index > x].prod())
    return d

def to_monthly(d, tol_days=550):
    d = d.sort_values("filed").rename(columns={"filed": "Date"})
    out = pd.merge_asof(pd.DataFrame({"Date": months}), d[["Date", "val"]],
                        on="Date", direction="backward",
                        tolerance=pd.Timedelta(days=tol_days))   # drop stale values
    return out.set_index("Date")["val"]

def edgar_series(t, sp):
    cik = cik_map.get(t) or cik_map.get(t.replace("-", "."))
    if cik is None:
        return None
    r = requests.get(f"https://data.sec.gov/api/xbrl/companyfacts/CIK{cik}.json", headers=H)
    if r.status_code != 200:
        return None
    facts, parts = r.json()["facts"], []
    for pri, (ns, tag) in enumerate(TAGS):
        try:
            d = pd.DataFrame(facts[ns][tag]["units"]["shares"])[["filed", "val"]]
        except KeyError:
            continue
        d["pri"] = pri
        parts.append(d)
    if not parts:
        return None
    d = pd.concat(parts)
    d["filed"] = pd.to_datetime(d["filed"]).astype("datetime64[ns]")
    d = d.sort_values(["filed", "pri"]).drop_duplicates("filed", keep="first")
    return to_monthly(split_adjust(d, sp))

def yahoo_series(t, sp):
    s = yf.Ticker(t).get_shares_full(start="2009-01-01", end="2025-01-31")
    if s is None or len(s) == 0:
        return None
    d = pd.DataFrame({"filed": s.index.tz_localize(None).normalize().astype("datetime64[ns]"),
                      "val": s.values})
    d = d.drop_duplicates("filed", keep="last")
    return to_monthly(split_adjust(d, sp))

# ---------- main loop ----------
shares, source = {}, {}
for i, t in enumerate(tickers):
    try:
        sp = splits_for(t)
        e = edgar_series(t, sp)
        y = yahoo_series(t, sp)
        if e is not None and y is not None:
            s = e.fillna(y); src = "EDGAR+Yahoo fill"
        elif e is not None:
            s = e; src = "EDGAR"
        elif y is not None:
            s = y; src = "Yahoo only"
        else:
            s = None; src = "none"
    except Exception as ex:
        s, src = None, f"error: {ex}"
    if s is not None:
        shares[t] = s
    source[t] = src
    time.sleep(0.2)                                  # stay under SEC's 10 requests/sec
    if i % 50 == 0:
        print(i, "of", len(tickers))

shares = pd.DataFrame(shares).reindex(columns=tickers)
# --- clean implausible share counts ---
shares = shares.where(shares > 0)                       # zeros -> NaN

# compare each month to the firm's own 5-year rolling median
med = shares.rolling(61, center=True, min_periods=12).median()
ratio = shares / med
bad = (ratio > 3) | (ratio < 1/3)                       # >3x away from its norm
print("share counts removed as outliers:", int(bad.sum().sum()))
print(bad.sum()[bad.sum() > 0])                         # which firms

shares = shares.mask(bad).ffill(limit=12)               # fill gaps with last good value
mcap = px.reindex(columns=tickers) * shares          # USD, split-consistent
pickle.dump((mcap, source), open("mcap.pkl", "wb"))

# ---------- coverage report ----------
cov = pd.DataFrame({
    "source": pd.Series(source),
    "first_mcap_date": mcap.apply(lambda c: c.first_valid_index()),
    "months_with_mcap": mcap.notna().sum(),
    "months_in_vol_data": vol[tickers].notna().sum(),
})
print(cov["source"].value_counts())
print("Firm-months with vol AND mcap:",
      (vol[tickers].notna() & mcap.notna()).sum().sum(), "of", vol[tickers].notna().sum().sum())

# ---------- Excel output ----------
bn = (mcap / 1e9).round(3)
with pd.ExcelWriter("market_cap.xlsx") as xw:
    bn.to_excel(xw, sheet_name="monthly_usd_bn")
    bn.loc[bn.index.month == 12].to_excel(xw, sheet_name="year_end_usd_bn")
    cov.to_excel(xw, sheet_name="coverage")
files.download("market_cap.xlsx")

# ---------- merge into your regression panel (lagged one month) ----------
size = np.log(mcap).shift(1)
sl = size.stack().rename("size").reset_index()
sl.columns = ["Date", "firm", "size"]
long["Date"] = pd.to_datetime(long["Date"]).astype("datetime64[ns]")
panel = long.merge(sl, on=["Date", "firm"], how="left")
print("Share of panel rows with size:", panel["size"].notna().mean().round(3))
```

```python
import requests, json, os, time, pickle
import pandas as pd, numpy as np

os.makedirs("edgar_cache", exist_ok=True)

def get_facts(t):
    cik = cik_map.get(t) or cik_map.get(t.replace("-", "."))
    if cik is None:
        return None
    path = f"edgar_cache/{cik}.json"
    if os.path.exists(path):
        return json.load(open(path))
    r = requests.get(f"https://data.sec.gov/api/xbrl/companyfacts/CIK{cik}.json", headers=H)
    time.sleep(0.2)                                   # SEC limit: 10 requests/sec
    if r.status_code != 200:
        return None
    j = r.json()
    json.dump(j, open(path, "w"))
    return j

def annual_series(facts, tags, flow):
    """One value per fiscal year from 10-K filings, in priority order of tags.
    Keeps the EARLIEST filing for each year-end (as first reported, no restatement look-ahead)."""
    parts = []
    for pri, tag in enumerate(tags):
        try:
            recs = facts["facts"]["us-gaap"][tag]["units"]["USD"]
        except KeyError:
            continue
        d = pd.DataFrame(recs)
        d = d[d["form"] == "10-K"].copy()
        if d.empty:
            continue
        d["end"] = pd.to_datetime(d["end"]); d["filed"] = pd.to_datetime(d["filed"])
        if flow:                                     # keep full-year flows only (not quarters / YTD)
            d["start"] = pd.to_datetime(d["start"])
            dur = (d["end"] - d["start"]).dt.days
            d = d[(dur >= 340) & (dur <= 380)]
        d["pri"] = pri
        parts.append(d[["end", "filed", "val", "pri"]])
    if not parts:
        return None
    d = pd.concat(parts).sort_values(["end", "pri", "filed"])
    d = d.drop_duplicates("end", keep="first")
    return d.set_index("end")[["val", "filed"]]

def firm_table(facts):
    A = annual_series(facts, ["Assets"], False)
    if A is None:
        return None
    df = A.rename(columns={"val": "assets", "filed": "f_assets"})
    pieces = {
        "tot": annual_series(facts, ["DebtLongtermAndShorttermCombinedAmount"], False),
        "cur": annual_series(facts, ["DebtCurrent", "LongTermDebtCurrent", "ShortTermBorrowings"], False),
        "nc":  annual_series(facts, ["LongTermDebtNoncurrent", "LongTermDebtAndCapitalLeaseObligations"], False),
        "ltd": annual_series(facts, ["LongTermDebt"], False),
        "cx":  annual_series(facts, ["PaymentsToAcquirePropertyPlantAndEquipment",
                                     "PaymentsToAcquireProductiveAssets"], True),
    }
    for k, s in pieces.items():
        if s is None:
            df[k] = np.nan; df["f_" + k] = pd.NaT
        else:
            df = df.join(s.rename(columns={"val": k, "filed": "f_" + k}))
    debt = df["tot"]
    debt = debt.fillna(df["nc"] + df["cur"].fillna(0))   # no current debt reported -> treated as 0
    debt = debt.fillna(df["ltd"])                         # last resort: LongTermDebt (incl. current part)
    out = pd.DataFrame({
        "leverage": debt / df["assets"],
        "capex_ta": df["cx"] / df["assets"],
        "filed": df[[c for c in df.columns if c.startswith("f_")]].max(axis=1),  # latest of components
    })
    out = out[(df["assets"] > 0)]
    return out.reset_index(drop=True)

def to_monthly(tab, col, months, tol_days=450):
    d = tab[["filed", col]].dropna().sort_values("filed").rename(columns={"filed": "Date"})
    if d.empty:
        return pd.Series(np.nan, index=months)
    d["Date"] = d["Date"].astype("datetime64[ns]")
    m = pd.DataFrame({"Date": months.astype("datetime64[ns]")})
    o = pd.merge_asof(m, d, on="Date", direction="backward",
                      tolerance=pd.Timedelta(days=tol_days))
    return pd.Series(o[col].values, index=months)

N = 10                                   # test with 10 firms; set to None for all
tickers = list(vol.columns)[:N] if N else list(vol.columns)
months = pd.DatetimeIndex(vol.index)

lev, cap, status = {}, {}, {}
for i, t in enumerate(tickers):
    try:
        f = get_facts(t)
        tab = firm_table(f) if f else None
        if tab is None or tab.empty:
            status[t] = "no data"
        else:
            lev[t] = to_monthly(tab, "leverage", months)
            cap[t] = to_monthly(tab, "capex_ta", months)
            status[t] = "ok"
    except Exception as ex:
        status[t] = f"error: {ex}"
    if i % 25 == 0:
        print(i, "of", len(tickers))

lev = pd.DataFrame(lev).reindex(columns=tickers).shift(1)   # extra one-month lag, like size
cap = pd.DataFrame(cap).reindex(columns=tickers).shift(1)
pickle.dump((lev, cap, status), open("lev_capex.pkl", "wb"))

print(pd.Series(status).value_counts())
print("months with leverage per firm (median):", lev.notna().sum().median())
print("months with capex/TA per firm (median):", cap.notna().sum().median())

def stack(df, name):
    s = df.stack().rename(name).reset_index()
    s.columns = ["Date", "firm", name]
    return s

base = panel if "panel" in globals() else long
base["Date"] = pd.to_datetime(base["Date"]).astype("datetime64[ns]")
panel = (base.drop(columns=[c for c in ["leverage", "capex_ta"] if c in base.columns])
             .merge(stack(lev, "leverage"), on=["Date", "firm"], how="left")
             .merge(stack(cap, "capex_ta"), on=["Date", "firm"], how="left"))

for c in ["leverage", "capex_ta"]:                           # winsorize at 1% / 99%
    lo, hi = panel[c].quantile([.01, .99])
    panel[c] = panel[c].clip(lo, hi)
print(panel[["leverage", "capex_ta"]].describe().round(3))
```
